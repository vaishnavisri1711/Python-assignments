### J. More Number Problems

```python
n = int(input("Enter a number: "))
if n < 2:
    print("Not prime")
else:
    factors = 0
    for d in range(1, n + 1):
        if n % d == 0:
            factors += 1
    if factors == 2:
        print("Prime")
    else:
        print("Not prime")
```

```python
limit = int(input("Limit: "))
even_sum = sum(range(2, limit + 1, 2))
odd_sum = sum(range(1, limit + 1, 2))
print("Sum of evens:", even_sum)
print("Sum of odds:", odd_sum)
```

```python
n = int(input("Factorial number: "))
if n < 0:
    print("Negative input is invalid")
else:
    result = 1
    for k in range(2, n + 1):
        result *= k
    print("Factorial =", result)
```
